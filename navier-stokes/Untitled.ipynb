```python
import xarray as xr

ds = xr.open_dataset("../../../fm_tutorial/dataset_gen/simu_pyqg_512_3.nc").sel(time=slice(60, 149))
print(ds.q.shape) 
print(ds)
```

```python
q = ds["q"].values 
q = q[70:, 0, :, :]    # shape (90, 512, 512)

time_lag = 2

z0_list = []
z1_list = []

for t in range(len(q) - time_lag):
    z0_list.append(q[t])
    z1_list.append(q[t + time_lag])

z0 = torch.tensor(np.stack(z0_list)).float()
z1 = torch.tensor(np.stack(z1_list)).float()
```

```python
print(z0.shape, z1.shape) 
```

```python
torch.save((z0, z1), "my_nse_512.pt")
```
