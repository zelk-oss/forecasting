```python
import xarray as xr
import numpy as np
```

```python
ds_train = xr.open_zarr(
    "data/wb2_train.zarr"
)["geopotential"].drop_vars("time")
```

```python
weights_lat = np.cos(np.deg2rad(ds_train.latitude))
weights_lat /= weights_lat.mean()
```

```python
weights_lat.values.round(6)
```

```python
ds_input = ds_train[:-1]
ds_residual = ds_train[1:]-ds_train[:-1]
```

# Input

```python
input_mean = (ds_input * weights_lat).mean(["time", "longitude", "latitude"]).compute()
input_std = np.sqrt((weights_lat * ((ds_input-input_mean))**2).mean(["time", "longitude", "latitude"])).compute()
```

```python
print(",".join([f"{v:.6f}" for v in input_mean.values]))
print(",".join([f"{v:.6f}" for v in input_std.values]))
```

# Residual

```python
residual_mean = (ds_residual * weights_lat).mean(["time", "longitude", "latitude"]).compute()
residual_std = np.sqrt((weights_lat * ((ds_residual-residual_mean))**2).mean(["time", "longitude", "latitude"])).compute()
```

```python
print(",".join([f"{v:.6f}" for v in residual_mean.values]))
print(",".join([f"{v:.6f}" for v in residual_std.values]))
```
