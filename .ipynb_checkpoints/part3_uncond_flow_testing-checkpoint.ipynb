```python
import torch

import xarray as xr
import numpy as np
import pandas as pd

from constants import *
from neural_net import get_net

from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy
import cmocean
```

```python
torch.manual_seed(10)
```

# Settings

```python
device = torch.device("cuda:1")
dtype = torch.bfloat16
```

```python
batch_size = 64

# Number of ensemble members
n_ensemble = 4

# Integration steps for flow model
n_int = 1

# 15 day forecast length
n_forecast_length = 10

lead_time_axis = pd.Timedelta("6h") * np.arange(1, n_forecast_length+1)
```

# Load data

```python
ds_test = xr.open_zarr("../data/sqg_test.zarr")["q"].compute()
```

# Estimate score for persistence

```python
initial_conditions = ds_test[:-n_forecast_length]
n_times_total = len(initial_conditions)
print(n_times_total) 
```

```python
persistence_mse = []

pbar = tqdm(np.arange(0, n_times_total, batch_size))
for start_idx in pbar:
    curr_initial = initial_conditions[start_idx:start_idx+batch_size]

    # Loop over lead times
    batch_mse = []
    for ld in range(1, n_forecast_length+1):
        # Get prediction
        curr_prediction = curr_initial.values

        # Error estimation
        error = curr_prediction - ds_test[start_idx+ld:start_idx+len(curr_prediction)+ld]
        mse = ((error**2)).mean(["longitude", "latitude"])
        mse = mse.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_mse.append(mse)
    batch_mse = xr.concat(batch_mse, dim="lead_time")
    persistence_mse.append(batch_mse)
persistence_mse = xr.concat(persistence_mse, dim="time")
```

# Deterministic forecast

## Load neural network

```python
n_blocks = 4
n_features = 64
det_model = get_net(
    n_input=1,
    n_output=1, n_blocks=n_blocks, n_features=n_features, mult=2,
    n_embedding=0, wave_length=0.1,
    device=device, dtype=dtype
)
state_dict = torch.load("../data/best_reg_model.ckpt", map_location="cpu")
det_model.load_state_dict(state_dict)
det_model = det_model.eval().to(device=device, dtype=dtype)
```

```python
det_mse = []

pbar = tqdm(np.arange(0, n_times_total, batch_size))
for start_idx in pbar:
    curr_initial = initial_conditions[start_idx:start_idx+batch_size]
    curr_prediction = curr_initial.values

    # Loop over lead times
    batch_mse = []
    for ld in tqdm(range(1, n_forecast_length+1), leave=False):
        # Get prediction
        net_io = torch.from_numpy(curr_prediction)
        net_io = ((net_io-in_mean) / in_std).to(device=device, dtype=dtype)
        with torch.no_grad():
            net_io = det_model(net_io).cpu().float()
        curr_prediction = curr_prediction + (net_io * res_std + res_mean).numpy()

        # Error estimation
        error = curr_prediction - ds_test[start_idx+ld:start_idx+len(curr_prediction)+ld]
        mse = (error**2).mean(["longitude", "latitude"])
        mse = mse.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_mse.append(mse)
    batch_mse = xr.concat(batch_mse, dim="lead_time")
    det_mse.append(batch_mse)
det_mse = xr.concat(det_mse, dim="time")
```

# Flow matching generation

```python
delta_t = 1 / n_int

def sample_with_flow(flow_model, n_samples, H, W):
    device = next(flow_model.parameters()).device
    dtype = next(flow_model.parameters()).dtype
    dynamics = torch.randn(n_samples, 1, H, W, device=device, dtype=dtype)
    for t in np.linspace(0, 1, n_int+1)[:-1]:
        pseudo_time = torch.full((n_samples, 1), t, device=device, dtype=dtype)
        velocity = flow_model(dynamics, pseudo_time)
        dynamics = dynamics + delta_t * velocity
    return dynamics
```

```python
ckpt = torch.load("../data/best_flowmodel_test.ckpt", map_location="cpu")
cfg = ckpt["config"]

flow_model = get_net(
    n_input=cfg["n_input"],
    n_output=cfg["n_output"],
    n_blocks=cfg["n_blocks"],
    n_features=cfg["n_features"],
    n_embedding=cfg["n_embedding"],
    wave_length=cfg["wave_length"],
    device=device,
    dtype=dtype,
)

# Load weights
flow_model.load_state_dict(ckpt["model_state_dict"])

# Switch to eval + move to device
flow_model = flow_model.eval().to(device=device, dtype=dtype)

# Optional: compile for speed
flow_model = torch.compile(flow_model, mode="reduce-overhead")
```

```python
flow_mse_ens = []
flow_mse_mean = []
flow_spread = []

# divide in batches and iterate over them 
# whta is a good number of batches to choose? 
pbar = tqdm(np.arange(0, n_times_total, batch_size))
for start_idx in pbar:
    curr_initial = initial_conditions[start_idx:start_idx+batch_size]
    curr_prediction = curr_initial.values

    # Add ensemble dimension
    curr_prediction = curr_prediction[None].repeat(n_ensemble, axis=0)

    # Loop over lead times
    batch_mse_ens = []
    batch_mse_mean = []
    batch_spread = []
    for ld in tqdm(range(1, n_forecast_length+1), leave=False):
        # curr_prediction shape: (n_ensemble, batch_size, 1, 512, 512)
        n_ens, n_batch, C, H, W = curr_prediction.shape
        with torch.no_grad():
            net_io = sample_with_flow(flow_model, n_samples=n_ens * n_batch, H=H, W=W)
            net_io = net_io.reshape(n_ens, n_batch, C, H, W).cpu().float()
        curr_prediction = curr_prediction + (net_io * res_std + res_mean).numpy()

        # Error estimation for ensemble members
        truth = ds_test[start_idx+ld:start_idx+curr_prediction.shape[1]+ld]
        truth = truth.expand_dims(ensemble=np.arange(n_ensemble), axis=0)
        error = curr_prediction - truth
        mse = (error**2).mean(["ensemble", "longitude", "latitude"])
        mse = mse.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_mse_ens.append(mse)

        # Error estimation for ensemble mean
        truth = ds_test[start_idx+ld:start_idx+curr_prediction.shape[1]+ld]
        error = curr_prediction.mean(axis=0) - truth
        mse = ((error**2)).mean(["longitude", "latitude"])
        mse = mse.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_mse_mean.append(mse)

        # Estimate spread as ensemble variance
        variance = curr_prediction.var(axis=0, ddof=1)
        variance = ds_test[start_idx+ld:start_idx+curr_prediction.shape[1]+ld].copy(data=variance)
        variance = (variance).mean(["longitude", "latitude"])
        variance = variance.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_spread.append(variance)

    batch_mse_ens = xr.concat(batch_mse_ens, dim="lead_time")
    flow_mse_ens.append(batch_mse_ens)

    batch_mse_mean = xr.concat(batch_mse_mean, dim="lead_time")
    flow_mse_mean.append(batch_mse_mean)

    batch_spread = xr.concat(batch_spread, dim="lead_time")
    flow_spread.append(batch_spread)

flow_mse_ens = xr.concat(flow_mse_ens, dim="time")
flow_mse_mean = xr.concat(flow_mse_mean, dim="time")
flow_spread = xr.concat(flow_spread, dim="time")
```

## Deterministic forecast 
Autoregressive residual forecast; at each step the model predicts the increment $\Delta q$, which is then added to the previous state. 

```python
for _ in tqdm(range(len(sliced_truth)-1)):
    net_io = torch.from_numpy(det_prediction[-1])
    net_io = ((net_io-in_mean) / in_std).to(device=device, dtype=dtype)
    with torch.no_grad():
        net_io = det_model(net_io).cpu().float()
    curr_prediction = det_prediction[-1] + (net_io * res_std + res_mean).numpy()
    det_prediction.append(curr_prediction)

det_prediction = np.concatenate(det_prediction, axis=0)
det_prediction = sliced_truth.copy(data=det_prediction)
```

## Flow model

```python
torch.manual_seed(42)
flow_prediction = [sliced_truth[0].values[None, None].repeat(n_ensemble, axis=0)]

for _ in tqdm(range(len(sliced_truth)-1)):
    last = flow_prediction[-1]  # shape: (n_ens, n_batch, C, H, W)
    n_ens, n_batch, C, H, W = last.shape
    with torch.no_grad():
        net_io = sample_with_flow(flow_model, n_samples=n_ens * n_batch, H=H, W=W)
        net_io = net_io.reshape(n_ens, n_batch, C, H, W).cpu().float()
    curr_prediction = flow_prediction[-1] + (net_io * res_std + res_mean).numpy()
    flow_prediction.append(curr_prediction)
flow_prediction = np.concatenate(flow_prediction, axis=1)
flow_prediction = sliced_truth.expand_dims(ensemble=np.arange(n_ensemble), axis=0).copy(data=flow_prediction)
```

```python
def plot_q(q, ax=None, title=''):
    if ax is None:
        fig, ax = plt.subplots()
    qmax = np.max(q) 
    pc = ax.pcolormesh(
        np.linspace(0, 2*np.pi, q.shape[1]),
        np.linspace(0, 2*np.pi, q.shape[0]),
        q, cmap='seismic', vmin=-qmax, vmax=qmax
    )
    ax.set_aspect(1)
    ax.set_title(title)
    return pc
```

```python
def plot_q(q, ax=None, title='', cbar=False):
    """Plot a 2D vorticity field. q should be shape (H, W)."""
    if ax is None:
        fig, ax = plt.subplots()
    q = np.asarray(q).squeeze()          # ensure 2-D
    assert q.ndim == 2, f"Expected 2D array, got shape {q.shape}"
    qmax = np.abs(q).max()
    pc = ax.pcolormesh(
        np.linspace(0, 2*np.pi, q.shape[1] + 1),   # (W+1,) cell edges
        np.linspace(0, 2*np.pi, q.shape[0] + 1),   # (H+1,) cell edges
        q, cmap='seismic', vmin=-qmax, vmax=qmax,
        shading='flat'
    )
    ax.set_aspect('equal')
    ax.set_title(title, fontsize=9)
    ax.set_xticks([]); ax.set_yticks([])
    return pc
```

## Generated spectrum 

```python
# ── Cell 2: Spectral comparison ───────────────────────────────────────────────

# physical parameters
f   = 1.5e-4
L   = 1100 * 1e3
Ro  = 0.011607
N   = 30 * f
U   = Ro * f * L

def calc_ispec(M, _var_dens, averaging=True, truncate=True, nfactor=1):
    """Isotropic spectrum from 2-D rfft2 power spectrum (M x M//2+1)."""
    var_dens = np.copy(_var_dens)
    var_dens[...,  0] /= 2
    var_dens[..., -1] /= 2

    ll = np.fft.fftfreq(M) * M
    kk = np.arange(0, M // 2 + 1)
    k2d, l2d = np.meshgrid(kk, ll)
    wv = np.sqrt(k2d**2 + l2d**2)

    kmax = np.minimum(np.abs(ll).max(), np.abs(kk).max()) if truncate \
           else np.sqrt(np.abs(ll).max()**2 + np.abs(kk).max()**2)
    dkr  = np.sqrt(2) * nfactor
    kr   = np.arange(0, kmax, dkr)
    phr  = np.zeros(kr.size)

    for i in range(kr.size):
        if i == kr.size - 1:
            mask = (wv >= kr[i]) & (wv <= kr[i] + dkr)
        else:
            mask = (wv >= kr[i]) & (wv < kr[i+1])
        phr[i] = (var_dens[mask].mean() * (kr[i] + dkr/2) * np.pi / 1
                  if averaging else var_dens[mask].sum() / dkr)
        phr[i] *= 2
    return kr + dkr/2, phr


def q_to_spec(q_2d, L=L, U=U):
    """
    q_2d : np.ndarray, shape (H, W) — single snapshot of PV.
    Returns (kr [m^-1], enstrophy_spectrum).
    """
    q_2d = np.asarray(q_2d).squeeze()
    assert q_2d.ndim == 2
    M = q_2d.shape[-1]
    vovo     = np.fft.rfft2(q_2d)
    ener_spec = np.abs(vovo)**2 / M**2          # (M, M//2+1)
    kr, enst  = calc_ispec(M, ener_spec)
    enst = enst * U**2 / L
    kr   = kr   / L
    return kr, enst


def mean_spectrum(q_array, L=L, U=U):
    """
    Average spectrum over a collection of snapshots.
    q_array: (N, H, W) or (N, 1, H, W)
    """
    q_array = np.asarray(q_array).squeeze()
    if q_array.ndim == 2:
        q_array = q_array[None]
    spectra = []
    for q in q_array:
        kr, s = q_to_spec(q, L=L, U=U)
        spectra.append(s)
    return kr, np.stack(spectra)


# ── gather spectra ─────────────────────────────────────────────────────────────

# ground truth — all time steps
truth_q = sliced_truth.values.squeeze()       # (T, H, W)
kr, truth_specs = mean_spectrum(truth_q)
truth_mean = truth_specs.mean(0)
truth_p25  = np.percentile(truth_specs, 25, axis=0)
truth_p75  = np.percentile(truth_specs, 75, axis=0)

# flow matching samples  (n_ens, T, C, H, W) — pool all members × times
fm_q = flow_prediction[:, :, 0].reshape(-1, *flow_prediction.shape[-2:])  # (N, H, W)
_,  fm_specs   = mean_spectrum(fm_q)
fm_mean  = fm_specs.mean(0)
fm_p25   = np.percentile(fm_specs, 25, axis=0)
fm_p75   = np.percentile(fm_specs, 75, axis=0)

# deterministic forecast — one member, all lead times
det_q = det_prediction.values.squeeze()       # (T, H, W) — adjust indexing
_, det_specs = mean_spectrum(det_q)
det_mean = det_specs.mean(0)

# ── plot ──────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 5), dpi=130)

ax.loglog(kr, truth_mean, color='steelblue',  lw=2.0, label='Truth (mean)')
ax.fill_between(kr, truth_p25, truth_p75, color='steelblue', alpha=0.20, label='Truth IQR')

ax.loglog(kr, fm_mean,    color='firebrick', lw=2.0, label='Flow matching (mean)')
ax.fill_between(kr, fm_p25, fm_p75,          color='firebrick', alpha=0.20, label='FM IQR')

ax.loglog(kr, det_mean,   color='black',     lw=1.5, ls='--', label='Deterministic')

# reference slopes
k_ref = np.array([1/(150e3), 1/(9e3)])
ax.loglog(k_ref, 3e-6 * k_ref**(-5/3), 'k:', lw=1, label='k⁻⁵/³')
ax.loglog(k_ref, 3e-6 * k_ref**(-3),   'k-.', lw=1, label='k⁻³')

ax.axvline(1/(19e3),  color='pink', ls='--', alpha=0.8, label='19 km')
ax.axvline(1/(100e3), color='pink', ls='--', alpha=0.8, label='100 km')
ax.set_xlim(1/(150e3), 1/(9e3))
ax.set_ylim(1e-7, 1e-2)
ax.set_xlabel("Wavenumber [m⁻¹]")
ax.set_ylabel("Enstrophy spectrum")
ax.set_title("PV enstrophy spectrum: truth vs flow matching vs deterministic")
ax.legend(fontsize=8, ncol=2)
plt.tight_layout()
plt.savefig("spectrum_comparison.png", dpi=130, bbox_inches='tight')
plt.show()
```

```python
# ── Cell 3: RMSE / spread / spread-skill ──────────────────────────────────────
# Assumes:
#   flow_mse_ens  : (time, var_name) — MSE of each ensemble member mean vs truth
#   flow_mse_mean : (time, var_name) — MSE of ensemble mean vs truth
#   flow_spread   : (time, var_name) — ensemble variance (spread²)
#   in_std        : normalisation std  (var_name,) tensor

std = in_std.squeeze().numpy()   # (C,) or scalar

def normalise(da):
    return (da / std**2).mean(["time", "var_name"])

rmse_ens  = np.sqrt(normalise(flow_mse_ens))
rmse_mean = np.sqrt(normalise(flow_mse_mean))
spread    = np.sqrt(normalise(flow_spread))

fig, axes = plt.subplots(1, 2, figsize=(12, 4), dpi=130)

# ── (a) nRMSE panel ───────────────────────────────────────────────────────────
ax = axes[0]
ax.plot(lead_time_axis / pd.Timedelta("1 day"),
        np.sqrt((persistence_mse / std**2).mean(["time", "var_name"])),
        label="Persistence", c="0.5", ls="--")
ax.plot(lead_time_axis / pd.Timedelta("1 day"),
        np.sqrt((det_mse / std**2).mean(["time", "var_name"])),
        label="Deterministic", c="black", ls="-")

# flow matching: since it's unconditional, summarise as a horizontal band
ax.axhline(float(rmse_mean), color='firebrick', lw=2,
           label=f"FM ensemble mean nRMSE = {float(rmse_mean):.3f}")
ax.axhspan(float(rmse_mean) - float(rmse_mean)*0.05,
           float(rmse_mean) + float(rmse_mean)*0.05,
           color='firebrick', alpha=0.15)
ax.axhline(float(rmse_ens), color='red', lw=1.5, ls=':',
           label=f"FM per-member nRMSE = {float(rmse_ens):.3f}")

ax.set_ylabel("nRMSE")
ax.set_xlabel("Lead time (days)")
ax.set_title("(a) Normalised RMSE vs lead time\n"
             "(FM shown as horizontal: no lead-time conditioning)")
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)

# ── (b) spread–skill ──────────────────────────────────────────────────────────
ax = axes[1]
ax.axhline(1.0, color='k', ls='--', lw=1, label='Perfect spread–skill')

# ideal: spread ≈ RMSE of ensemble mean  → ratio ~1
ratio = float(spread) / float(rmse_mean)
ax.bar(['FM'], [ratio], color='firebrick', alpha=0.7, width=0.3)
ax.set_ylim(0, max(2.0, ratio * 1.3))
ax.set_ylabel("Spread / RMSE (ensemble mean)")
ax.set_title("(b) Spread–skill ratio\n(>1 = over-dispersive, <1 = under-dispersive)")
ax.legend()
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig("rmse_spread.png", dpi=130, bbox_inches='tight')
plt.show()
```

## Distribution comparison

```python
# ── Cell 4: Marginal distribution & Q-Q plot ─────────────────────────────────
truth_flat = sliced_truth.values.flatten()
fm_flat    = flow_prediction[:, :, 0].flatten()   # channel 0

fig, axes = plt.subplots(1, 2, figsize=(11, 4), dpi=130)

# ── PDF ───────────────────────────────────────────────────────────────────────
ax = axes[0]
bins = np.linspace(
    min(truth_flat.min(), fm_flat.min()),
    max(truth_flat.max(), fm_flat.max()),
    80
)
ax.hist(truth_flat, bins=bins, density=True, alpha=0.5,
        color='steelblue', label='Truth')
ax.hist(fm_flat,    bins=bins, density=True, alpha=0.5,
        color='firebrick', label='Flow matching')
ax.set_xlabel("q (PV)")
ax.set_ylabel("Density")
ax.set_title("Marginal PDF of PV")
ax.legend()
ax.grid(True, alpha=0.3)

# ── Q-Q plot ─────────────────────────────────────────────────────────────────
ax = axes[1]
quantiles = np.linspace(0.01, 0.99, 200)
q_truth = np.quantile(truth_flat, quantiles)
q_fm    = np.quantile(fm_flat,    quantiles)
ax.scatter(q_truth, q_fm, s=8, c='firebrick', alpha=0.6, label='FM vs Truth')
lims = [min(q_truth.min(), q_fm.min()), max(q_truth.max(), q_fm.max())]
ax.plot(lims, lims, 'k--', lw=1, label='y = x (perfect)')
ax.set_xlabel("Truth quantiles")
ax.set_ylabel("FM quantiles")
ax.set_title("Q–Q plot: Flow matching vs truth")
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("distribution_comparison.png", dpi=130, bbox_inches='tight')
plt.show()
```
