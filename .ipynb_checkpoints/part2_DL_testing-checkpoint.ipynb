```python
batch_size = 8

# 15 day forecast length
n_forecast_length = 5

lead_time_axis = pd.Timedelta("6h") * np.arange(1, n_forecast_length+1)
```

# Load data

```python
ds_test = xr.open_zarr("../data/sqg_test.zarr")["q"].compute()
```

# Estimate score for persistence

```python
initial_conditions = ds_test[:-n_forecast_length]
n_times_total = len(initial_conditions)
print(n_times_total)
print(n_forecast_length)
```

```python
persistence_mse = [] # mse when state[t+1]=state[t] for every t 
# persistance is the baseline 

pbar = tqdm(np.arange(0, n_times_total, batch_size))
for start_idx in pbar:
    curr_initial = initial_conditions[start_idx:start_idx+batch_size]

    # Loop over lead times
    batch_mse = []
    for ld in range(1, n_forecast_length+1):
        # Get prediction
        curr_prediction = curr_initial.values # persistance forecast
        # for every lead time = 6hrs --> same conditions 

        # Error estimation --> comparing persistence forecast at time t with actual ERA5 data for that time  
        error = curr_prediction - ds_test[start_idx+ld:start_idx+len(curr_prediction)+ld]
        # get error and average over spatial dimensions to get MSE per time sample (starting time) and lead time (delta t) 
        mse = ((error**2) * weights_lat.numpy()).mean(["longitude", "latitude"]) 
        mse = mse.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_mse.append(mse)
    batch_mse = xr.concat(batch_mse, dim="lead_time")
    persistence_mse.append(batch_mse)
# concatenate all computed MSE values 
persistence_mse = xr.concat(persistence_mse, dim="time")
```

```python
persistence_mse.shape # (forecast length, time, variables) 
# time = total time steps - forecast length for horizon
```

```python
mse_by_lead_time = persistence_mse.mean(dim="time")
print(mse_by_lead_time.shape)

plt.figure(figsize=(12,6))

plt.plot(mse_by_lead_time[:, 0])
plt.title(f"q")
plt.ylabel("MSE")

plt.xlabel("Lead time index")
plt.tight_layout()
plt.show()
```

# Loop neural network to get NN forecast

## Load neural network

```python
# perhaps you forgot which hyperparameters the trained net had... 
sd = torch.load("../data/best_detmodel_test.ckpt", map_location="cpu")
for k, v in sd.items():
    print(k, type(v))

for k, v in sd["model_state_dict"].items():
    print(k, tuple(v.shape))
```

```python
ckpt = torch.load("../data/best_detmodel_test.ckpt", map_location="cpu")

net = get_net(**ckpt["config"], device=device, dtype=dtype)

net.load_state_dict(ckpt["model_state_dict"])
net = net.eval().to(device=device, dtype=dtype)
```

```python
downsample_factor = ckpt["config"]["token_downsample_factor"]
print(f"Model internal downsample factor: {downsample_factor}")

all_data = initial_conditions  # shape: (n_times_total, 1, H, W)

# Rollout loop
nn_mse = []
pbar = tqdm(np.arange(0, n_times_total - n_forecast_length, batch_size))

for start_idx in pbar:
    batch_end_idx = min(start_idx + batch_size, n_times_total - n_forecast_length)
    actual_batch_size = batch_end_idx - start_idx
    
    # Batch of initial conditions (B, 1, H, W) - FULL resolution
    curr_state_fullres = torch.from_numpy(
        all_data.isel(time=slice(start_idx, batch_end_idx)).values
    )  # (B, 1, 256, 256)
    batch_mse = []
    batch_time_coords = all_data.isel(time=slice(start_idx, batch_end_idx)).time.values
    
    for ld in tqdm(range(1, n_forecast_length + 1), leave=False):
        # Normalize input (FULL resolution)
        net_in = ((curr_state_fullres - in_mean) / in_std).to(device=device, dtype=dtype)
        # Predict residual - output è downsampled
        with torch.no_grad():
            pred_residual_norm = net(net_in).cpu().float()  # (B, 1, H/factor, W/factor)
        # Denormalize predicted residual
        pred_residual = pred_residual_norm * res_std + res_mean
        
        # Get ground truth for next timestep (full resolution)
        gt_next_fullres = torch.from_numpy(
            all_data.isel(time=slice(start_idx + ld, batch_end_idx + ld)).values
        )  # (B, 1, 256, 256)
        
        # Downsample ground truth alla risoluzione del modello
        gt_next_modelres = torch.nn.functional.avg_pool2d(
            gt_next_fullres, 
            kernel_size=downsample_factor
        )  # (B, 1, 32, 32)
        
        # Downsample current state alla risoluzione del modello
        curr_state_modelres = torch.nn.functional.avg_pool2d(
            curr_state_fullres,
            kernel_size=downsample_factor
        )  # (B, 1, 32, 32)
        
        # Compute ground truth residual at model resolution
        true_residual = gt_next_modelres - curr_state_modelres
        
        # Compute MSE
        error = pred_residual - true_residual
        mse = (error**2).mean(dim=(1, 2, 3))  # (B,)
        
        # Store as xarray DataArray
        mse_da = xr.DataArray(
            mse.numpy(),
            dims=["time"],
            coords={"time": batch_time_coords}
        ).expand_dims(lead_time=[ld * pd.Timedelta("6h")])
        batch_mse.append(mse_da)
        # Update current state: upsample predicted residual
        pred_residual_upsampled = torch.nn.functional.interpolate(
            pred_residual,
            scale_factor=downsample_factor,
            mode='bilinear',
            align_corners=False
        )  # (B, 1, 256, 256)
        curr_state_fullres = curr_state_fullres + pred_residual_upsampled
    
    # Concatenate batch along lead_time
    batch_mse = xr.concat(batch_mse, dim="lead_time")
    nn_mse.append(batch_mse)

# Concatenate all batches along time
nn_mse = xr.concat(nn_mse, dim="time")
```

```python
print(nn_mse.shape) # same shape as the persistance mse 
```

```python
nn_mse_by_lead_time = nn_mse.mean(dim="time")
print(nn_mse_by_lead_time.shape)

plt.figure(figsize=(12, 6)) 

plt.plot(nn_mse_by_lead_time[:], label="nn MSE")
plt.plot(mse_by_lead_time[:], label="persistance MSE")
plt.title("q")
plt.ylabel("nn MSE")
plt.legend()

plt.xlabel("Lead time index")
plt.tight_layout()
plt.show()
```

Clearly different fields have different natural variability, therefore the MSE varies too over very different scales. To have comparable values we normalize by input standard deviation --> rMSE

```python
def to_tensor(x, device=device, dtype=dtype):
    if isinstance(x, np.ndarray):
        return torch.from_numpy(x).to(device=device, dtype=dtype)
    elif isinstance(x, torch.Tensor):
        return x.to(device=device, dtype=dtype)
    else:
        raise TypeError(f"Unsupported type {type(x)}")

in_mean  = to_tensor(in_mean)
in_std   = to_tensor(in_std)
res_mean = to_tensor(res_mean)
res_std  = to_tensor(res_std)

current = sliced_truth.values[0][None]  # (1,1,512,512)
current = torch.from_numpy(current).to(device=device, dtype=dtype)
predictions = [current.cpu().to(torch.float32).numpy()]

for _ in tqdm(range(len(sliced_truth) - 1)):

    net_io = (current - in_mean) / in_std   # normalize
    with torch.no_grad():
        net_out = net(net_io)  # your model prediction

        # downsample if needed
        downsampled = torch.nn.functional.avg_pool2d(current, kernel_size=downsample_factor)

        # add residual back in original resolution
        downsampled = downsampled + (net_out * res_std + res_mean)

        predictions.append(current.cpu().to(torch.float32).numpy())

prediction = np.concatenate(predictions, axis=0)
prediction = sliced_truth.copy(data=prediction)
```

```python
def plot_q(ax, q, qmax=None):
    q2d = q.squeeze()
    n = q2d.shape[0]
    x = np.linspace(0, 2*np.pi, n)
    y = np.linspace(0, 2*np.pi, n)
    pc = ax.pcolormesh(x, y, q2d, cmap="seismic", shading="auto")
    if qmax is None:
        qmax = np.abs(q2d).max()
    pc.set_clim(-qmax, qmax)
    ax.set_aspect("equal")
    ax.set_xlim(0, 2*np.pi)
    ax.set_ylim(0, 2*np.pi)

    return pc
```

```python
def plot_predictions_grid(prediction, step=1):
    """
    Plot prediction fields in rows of 3.
    step = skip frames to avoid 100 tiny plots.
    """
    fields = prediction[::step]
    n = len(fields)

    ncols = 3
    nrows = int(np.ceil(n / ncols))

    fig, axes = plt.subplots(nrows, ncols, figsize=(4*ncols, 4*nrows))

    # Flatten axes for easy looping
    axes = np.atleast_1d(axes).ravel()

    # global symmetric color scale
    qmax = np.abs(fields).max()

    for i, (ax, q) in enumerate(zip(axes, fields)):
        pc = plot_q(ax, q, qmax=qmax)
        ax.set_title(f"t = {i*step}")

    # Hide unused panels
    for ax in axes[n:]:
        ax.axis("off")

    # One shared colorbar
    fig.colorbar(pc, ax=axes[:n], shrink=0.8)

    plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6,6))

# ensure q is a 2D NumPy array
q_to_plot = (prediction[0] - prediction[1])
if isinstance(q_to_plot, torch.Tensor):
    q_to_plot = q_to_plot.squeeze().cpu().to(torch.float32).numpy()
else:
    q_to_plot = q_to_plot.squeeze().astype(np.float32)

# plot
pc = plot_q(ax, q_to_plot)

# pass the QuadMesh to colorbar
plt.colorbar(pc, ax=ax)

plt.show()
```
