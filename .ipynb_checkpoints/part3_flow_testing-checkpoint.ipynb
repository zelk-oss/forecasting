```python
# ============================================================================
# Helper functions for shape handling via downsampling/upsampling
# ============================================================================

def downsample(x, factor):
    """
    Downsample tensor/array by factor using bilinear interpolation.
    Assumes x is numpy or torch tensor with shape [..., C, H, W].
    """
    if factor == 1:
        return x
    
    # Convert to torch if needed
    is_numpy = isinstance(x, np.ndarray)
    if is_numpy:
        x_torch = torch.from_numpy(x).float()
    else:
        x_torch = x.float()
    
    # Reshape to [batch_eff, C, H, W]
    original_shape = x_torch.shape[:-2]
    C, H, W = x_torch.shape[-3:]
    x_flat = x_torch.reshape(-1, C, H, W)
    
    # Downsample
    new_H, new_W = H // factor, W // factor
    x_down = torch.nn.functional.interpolate(
        x_flat, size=(new_H, new_W), mode='bilinear', align_corners=False
    )
    
    # Reshape back
    x_result = x_down.reshape(*original_shape, C, new_H, new_W)
    
    return x_result.numpy() if is_numpy else x_result


def upsample(x, factor, target_size):
    """
    Upsample tensor/array back to target_size using bilinear interpolation.
    target_size should be (H, W).
    """
    if factor == 1:
        return x
    
    # Convert to torch if needed
    is_numpy = isinstance(x, np.ndarray)
    if is_numpy:
        x_torch = torch.from_numpy(x).float()
    else:
        x_torch = x.float()
    
    # Reshape to [batch_eff, C, H, W]
    original_shape = x_torch.shape[:-2]
    C = x_torch.shape[-3]
    x_flat = x_torch.reshape(-1, C, x_torch.shape[-2], x_torch.shape[-1])
    
    # Upsample
    x_up = torch.nn.functional.interpolate(
        x_flat, size=target_size, mode='bilinear', align_corners=False
    )
    
    # Reshape back
    H_target, W_target = target_size
    x_result = x_up.reshape(*original_shape, C, H_target, W_target)
    
    return x_result.numpy() if is_numpy else x_result
```

# Flow matching forecasts

### Numerical integration of a learned velocity field to transform noise to data sample conditioned on the initial state
- starts from random noise
- integrates in pseudo-time $t \in [0,1] $ (Euler integration of ODE) 
- uses the flow-matching neural velocity field
- ends with a generated sample consistent with the conditioning state.

```python
delta_t = 1 / n_int # integration step size 

def sample_with_flow(flow_model, initial_conditions, n_integration_steps=None):
    """
    Sample from flow model by integrating from noise to data.
    
    Args:
        flow_model: Trained flow matching model expecting (z_t, initial_cond, pseudo_time)
        initial_conditions: Initial conditions [ensemble, batch, channels, H, W]
        n_integration_steps: Number of integration steps (uses global n_int if None)
    
    Returns:
        Samples with same shape as initial_conditions
    """
    if n_integration_steps is None:
        n_integration_steps = n_int
    
    # Reshape to combine ensemble and batch dimensions
    n_ensemble_local = initial_conditions.shape[0]
    n_batch_local = initial_conditions.shape[1]
    # merge ensemble and batch dimension [ ensemble, batch, C, H, W] --> [ensemble x batch, C, H, W] 
    reshaped_conditions = torch.reshape(initial_conditions, (-1, *initial_conditions.shape[2:]))
    
    # Initialize dynamics with random noise
    # z0 ~ N(0,I)
    dynamics = torch.randn_like(reshaped_conditions)
    
    # Integration loop over pseudo-time t ∈ [0, 1)
    delta_t_local = 1 / n_integration_steps
    for t in np.linspace(0, 1, n_integration_steps + 1)[:-1]:
        # Construct pseudo time vector [batch_size, 1]
        pseudo_time = torch.full(
            (reshaped_conditions.shape[0], 1), t,
            device=dynamics.device, dtype=dynamics.dtype
        )
        
        # Concatenate intermediate state and initial conditions
        # Model expects [intermediate_state, initial_condition] as input
        in_tensor = torch.cat((dynamics, reshaped_conditions), dim=1)
        
        # Get velocity prediction from model
        with torch.no_grad():
            velocity = flow_model(in_tensor, pseudo_time)
        
        # Euler step: z_{t + Δt} = z_t + Δt * v_θ(z_t, t)
        dynamics = dynamics + delta_t_local * velocity
    
    # Reshape back to original dimensions [ensemble, batch, channels, H, W]
    dynamics = torch.reshape(dynamics, initial_conditions.shape)
    return dynamics
```

### Probabilistic forecast evaluation
Evaluate ensemble forecast, ensemble mean, ensemble spread 

```python
print("\n" + "=" * 70)
print("FLOW MATCHING FORECAST EVALUATION (physical space)")
print("=" * 70)

flow_mse_ens = []
flow_mse_mean = []
flow_spread = []

pbar = tqdm(np.arange(0, n_times_total, batch_size))
# select several starting states and forecast them forward through the model 
for start_idx in pbar:
    curr_initial = initial_conditions[start_idx:start_idx+batch_size]
    curr_prediction = curr_initial.values  # [batch, C, H, W] in physical space

    # Loop over lead times
    batch_mse_ens = []
    batch_mse_mean = []
    batch_spread = []
    for ld in tqdm(range(1, n_forecast_length+1), leave=False):
        # Downsample to model space and add ensemble dimension
        pred_model = downsample(curr_prediction, flow_downsample_factor)
        pred_model_ens = pred_model[None].repeat(n_ensemble, axis=0)  # [ensemble, batch, C, H, W]
        
        # Normalize for flow model
        net_io = torch.from_numpy(pred_model_ens)
        net_io = ((net_io-in_mean) / in_std).to(device=device, dtype=dtype)
        
        # Sample stochastic increments from flow
        with torch.no_grad():
            vel_model = sample_with_flow(flow_model, net_io).cpu().float()
        
        # Denormalize in model space
        vel_model = (vel_model * res_std + res_mean).numpy()  # [ensemble, batch, C, H_model, W_model]
        
        # Upsample residuals back to physical space
        phys_size = (curr_prediction.shape[-2], curr_prediction.shape[-1])
        vel_phys = upsample(vel_model, flow_downsample_factor, phys_size)  # [ensemble, batch, C, H, W]
        
        # Update prediction for all ensemble members
        curr_prediction_ens = curr_prediction[None] + vel_phys
        
        # Use ensemble mean for next iteration
        curr_prediction = curr_prediction_ens.mean(axis=0)

        # Error estimation for ensemble members
        truth = ds_test[start_idx+ld:start_idx+curr_prediction_ens.shape[1]+ld]
        truth_ens = truth.expand_dims(ensemble=np.arange(n_ensemble), axis=0)
        
        error_ens = curr_prediction_ens - truth_ens.values  # [ensemble, batch, C, H, W]
        mse_ens = ((error_ens**2) * weights_lat.numpy()).mean(["ensemble", "longitude", "latitude"])
        mse_ens = mse_ens.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_mse_ens.append(mse_ens)

        # Error estimation for ensemble mean
        truth = ds_test[start_idx+ld:start_idx+curr_prediction.shape[0]+ld]
        error_mean = curr_prediction - truth.values  # [batch, C, H, W]
        mse_mean = ((error_mean**2) * weights_lat.numpy()).mean(["longitude", "latitude"])
        mse_mean = mse_mean.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_mse_mean.append(mse_mean)

        # Estimate spread as ensemble variance
        variance = curr_prediction_ens.var(axis=0, ddof=1)
        variance_da = ds_test[start_idx+ld:start_idx+curr_prediction.shape[0]+ld].copy(data=variance)
        variance_weighted = (variance_da * weights_lat.numpy()).mean(["longitude", "latitude"])
        variance_weighted = variance_weighted.expand_dims(lead_time=[ld*pd.Timedelta("6h")]).assign_coords(time=curr_initial.indexes["time"])
        batch_spread.append(variance_weighted)

    batch_mse_ens = xr.concat(batch_mse_ens, dim="lead_time")
    flow_mse_ens.append(batch_mse_ens)

    batch_mse_mean = xr.concat(batch_mse_mean, dim="lead_time")
    flow_mse_mean.append(batch_mse_mean)

    batch_spread = xr.concat(batch_spread, dim="lead_time")
    flow_spread.append(batch_spread)

flow_mse_ens = xr.concat(flow_mse_ens, dim="time")
flow_mse_mean = xr.concat(flow_mse_mean, dim="time")
flow_spread = xr.concat(flow_spread, dim="time")
print(f"✓ Flow forecast complete. Shapes: ens={flow_mse_ens.shape}, mean={flow_mse_mean.shape}, spread={flow_spread.shape}")
```

```python
# check that det model and flow model are compatible (same dimension idk) 
def compare_velocity_and_residual(det_model, flow_model, x, t=0.99):
    """
    Compare deterministic residual with flow velocity near t=1.
    
    x: tensor [batch, C, H, W] (real data state)
    """
    device = next(det_model.parameters()).device
    x = x.to(device)

    # --- deterministic residual ---
    with torch.no_grad():
        det_res = det_model(x)

    # --- flow velocity ---
    pseudo_time = torch.full((x.shape[0], 1), t, device=device, dtype=x.dtype)

    # flow model expects concatenated [z_t, condition]
    flow_in = torch.cat([x, x], dim=1)

    with torch.no_grad():
        flow_vel = flow_model(flow_in, pseudo_time)

    # --- compare ---
    mse = torch.mean((det_res - flow_vel) ** 2).item()
    corr = torch.nn.functional.cosine_similarity(
        det_res.flatten(1), flow_vel.flatten(1), dim=1
    ).mean().item()

    print(f"MSE(det residual vs flow velocity) = {mse:.3e}")
    print(f"Cosine similarity = {corr:.3f}")

    return mse, corr

compare_velocity_and_residual(det_model, flow_model) 
```
