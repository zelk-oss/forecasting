```python
import json
import matplotlib.pyplot as plt

# Load loss history
with open("loss_history.json", "r") as f:
    loss_history = json.load(f)

train_losses = loss_history["train"]
val_losses   = loss_history["val"]

# Recreate epoch index
epochs = range(1, len(train_losses) + 1)

# Plot
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(epochs[100:], train_losses[100:], label="Train MSE", marker="o", markersize=3)
ax.plot(epochs[100:], val_losses[100:],   label="Val MSE",   marker="o", markersize=3)

ax.set_xlabel("Epoch")
ax.set_ylabel("MSE")
ax.set_title("Training & Validation Loss")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
